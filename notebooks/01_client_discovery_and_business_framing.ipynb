# Client Discovery and Business Framing

## 1. Business Context

The client has experienced an increase in third-party fraud (TPF) losses over the past several months despite relatively stable application volume. Historically, confirmed TPF represented approximately 1.8% of approved applications, compared with approximately 2.5–2.7% in recent months.

The current fraud strategy uses a third-party fraud score ranging from 0–999, where higher scores indicate greater fraud risk. Applications scoring 700 or higher are routed to manual review, while applications below 700 generally continue through the standard decision process, assuming they satisfy other identity and credit requirements.

The client's concern is not limited to increasing fraud losses. Manual review creates operational cost and can introduce friction and delays for legitimate applicants. Leadership therefore wants to understand whether the current strategy appropriately balances fraud prevention, operational cost, and customer experience rather than simply minimizing fraud at all costs.

The objective of this engagement is to evaluate the performance of the current fraud strategy, identify opportunities for improvement, and quantify the tradeoffs associated with potential strategy changes.


## 2. Current Application Decision Flow

Discovery established the following simplified application funnel:

Application
→ Eligibility / Completeness
→ KYC / CIP
→ Fraud Score & Fraud Strategy
→ Credit Underwriting
→ Booking
→ Matured Outcomes

Approximately 94% of applications passing initial eligibility/completeness checks pass KYC/CIP and reach the fraud-scoring stage.

At the fraud-strategy stage:

- Applications with a fraud score below 700 generally continue through the standard process.
- Applications scoring 700 or higher are routed to manual review.
- Reviewers may clear the application, decline it for suspected fraud, or require additional verification.
- Available step-up treatments can include document verification, phone/OTP verification, email verification, and additional identity verification.
- Fraud review must be resolved before the application proceeds to credit underwriting.
- Credit underwriting can independently decline applications that have cleared fraud controls.
- Only booked accounts subsequently generate observable post-booking performance outcomes.

This sequencing is important because not every scored application ultimately produces an observable fraud outcome.


## 3. Primary Business Questions

The analysis will seek to answer the following questions:

1. How effectively does the current fraud score separate confirmed third-party fraud from legitimate applicants?

2. Where are false positives and false negatives occurring under the current strategy?

3. Is the current 700 manual-review threshold appropriately balancing fraud detection, operational cost, and legitimate-customer friction?

4. How effective is manual review, particularly for high-risk applications that are reviewed and subsequently released?

5. Are there score ranges where alternative treatments may be more appropriate, such as standard approval, step-up verification, manual review, or potentially automatic decline?

6. How would alternative score thresholds or treatment strategies affect fraud losses, review volume, treatment costs, and the number of legitimate applicants affected?

7. Does fraud performance vary meaningfully over time or across important application segments such as channel?


## 4. Evaluation Population and Outcome Definition

The primary fraud outcome for this engagement is confirmed third-party fraud / identity theft, defined as an application made using another person's identity or identity information without authorization.

The strongest outcome evidence includes verified identity-theft claims from victims and post-booking investigations that establish sufficient evidence that the applicant was not the legitimate identity owner.

First-party fraud, ordinary credit losses, and early payment delinquency will not automatically be classified as confirmed TPF.

Outcome maturity must also be considered. Fraud outcomes are not immediately observable after booking. The analysis will generally require at least 90 days of outcome maturity, with 180 days preferred for the primary fully matured evaluation population.

An application without a confirmed fraud label should not automatically be interpreted as known legitimate if sufficient outcome time has not elapsed.

Applications declined or otherwise stopped for suspected fraud before booking also cannot be treated as confirmed fraud simply because the strategy identified them as suspicious. Their post-booking counterfactual outcome is unobservable.

The primary score-performance analysis will therefore focus on populations with sufficiently mature and observable outcomes. More recent vintages may be evaluated separately for directional trends where appropriate.


## 5. Current Strategy Observations from Discovery

Several observations from discovery will guide the analysis:

- Approximately 8% of recently scored applications are routed to manual review under the current 700 cutoff.
- Approximately 60% of recent confirmed fraud scored at or above 700 at application.
- Approximately 40% of recent confirmed fraud occurred below the current 700 threshold.
- Some applications scoring above 700 were manually reviewed, released by reviewers, booked, and later confirmed as fraud.
- The effectiveness of manual review across different score ranges has not been rigorously evaluated.
- The recent increase in fraud appears to have developed steadily rather than being driven by a single isolated spike.
- No major application-flow, marketing, or intentional customer-segment changes have been identified, although changes in applicant composition have not yet been ruled out.

These observations are hypotheses and starting points for analysis rather than predetermined conclusions.


## 6. Economic Considerations

The following approximate costs were established during discovery:

- Average realized loss per confirmed TPF account: ~$3,000
- Fully loaded manual-review cost: ~$12 per review
- Step-up verification vendor cost: ~$3 per application

The financial impact of incorrectly rejecting a legitimate applicant is not sufficiently known to assign a reliable dollar value. Therefore, legitimate-customer impact will be reported using counts and rates alongside financial estimates rather than assigning an unsupported monetary value.

Potential strategy changes should therefore be evaluated across multiple dimensions, including:

- Confirmed fraud captured
- Fraud losses potentially avoided
- Manual-review volume and cost
- Step-up verification volume and cost
- Legitimate applicants affected
- Overall application/customer friction


## 7. Scope

### Primary Scope

The primary engagement will evaluate the current fraud strategy and identify potential opportunities to improve the balance between fraud prevention, operational cost, and legitimate-customer experience.

The analysis will include:

- Fraud-score performance
- Performance over time
- Performance across relevant segments
- Current threshold effectiveness
- Manual-review effectiveness
- False-positive and false-negative tradeoffs
- Alternative threshold and treatment scenarios
- Financial and operational impact of potential strategy changes

### Future / Secondary Opportunities

Discovery also identified several potentially valuable analyses that are outside the primary scope of this engagement:

- Investigating whether early-payment delinquency may identify additional missed fraud.
- Evaluating whether fraud scoring could occur earlier in the application funnel before KYC/CIP.
- Retrospectively scoring populations that currently fail KYC/CIP and therefore never receive a fraud score.
- Comparing the current fraud solution with future model or product versions.

These opportunities may be explored separately after the current fraud strategy has been evaluated.
